```python
from scipy.stats import mannwhitneyu

results = []

for feature in numeric_features:
    no_group = df[df["Attrition"] == "No"][feature]
    yes_group = df[df["Attrition"] == "Yes"][feature]

    stat, p_value = mannwhitneyu(
        no_group,
        yes_group,
        alternative="two-sided"
    )

    results.append({
        "Feature": feature,
        "U_Statistic": stat,
        "P_Value": p_value
    })

mann_whitney_results = pd.DataFrame(results)

mann_whitney_results
```

None of the numerical features showed a statistically significant difference between the two attrition groups according to the Mann–Whitney U test.

### Observation

- The Mann–Whitney U test was performed to compare numerical features between employees with `Attrition = No` and `Attrition = Yes`.
- Using a significance level of 0.05, none of the numerical features has a p-value below 0.05.
- Therefore, no statistically significant difference was found between the two attrition groups for the numerical features based on this test.
- `Work_Life_Balance` and `Job_Involvement` have relatively lower p-values, but they are still above the 0.05 significance threshold.
- Statistical significance alone will not be used for final feature selection; model-based importance and domain relevance will also be considered.

```python
overtime_attrition = pd.crosstab(
    df["Overtime"],
    df["Attrition"],
    normalize="index"
) * 100

print(overtime_attrition)
```

```python
from scipy.stats import chi2_contingency
contingency_table = pd.crosstab(
    df["Overtime"],
    df["Attrition"]
)

print(contingency_table)
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2)
print("p-value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
from scipy.stats import chi2_contingency

chi_square_results = []

for feature in categorical_features:
    
    contingency_table = pd.crosstab(
        df[feature],
        df["Attrition"]
    )
    
    chi2, p_value, dof, expected = chi2_contingency(contingency_table)
    
    chi_square_results.append({
        "Feature": feature,
        "Chi2_Statistic": chi2,
        "P_Value": p_value,
        "Degrees_of_Freedom": dof
    })

chi_square_results = pd.DataFrame(chi_square_results)

chi_square_results
```

### Observation

- The Chi-Square test was performed to examine the association between categorical features and `Attrition`.
- `Gender`, `Marital_Status`, `Department`, and `Overtime` have p-values greater than 0.05, indicating no statistically significant association with attrition.
- `Job_Role` has a p-value of 0.028995, which is below the 0.05 significance level.
- Therefore, `Job_Role` shows a statistically significant association with employee attrition in this dataset.
- However, statistical significance does not indicate causation or measure the strength of the association. Further analysis will be performed before final feature selection.

# feature engineering

```python
df_fe = df.copy()

# Drop constant and identifier columns
df_fe.drop(
    columns=["EmployeeCount", "Over18", "StandardHours", "EmployeeNumber"],
    inplace=True
)
```

```python
df_fe["YearsOutsideCompany"] = (
    df_fe["TotalWorkingYears"] - df_fe["YearsAtCompany"]
)
```

```python
df_fe["CareerStagnation"] = (
    (df_fe["YearsSinceLastPromotion"] >= 5) &
    (df_fe["YearsInCurrentRole"] >= 5)
).astype(int)
```

```python
df_fe[["YearsOutsideCompany", "CareerStagnation"]].describe()
```

```python
df_fe.groupby("Attrition")[
    ["YearsOutsideCompany", "CareerStagnation"]
].mean()
```

```python
df_fe.drop(columns=["CareerStagnation"], inplace=True)
```

```python
df_fe["Overtime_LowSatisfaction"] = (
    (df_fe["OverTime"] == "Yes") &
    (df_fe["JobSatisfaction"] <= 2)
).astype(int)
```

```python
pd.crosstab(
    df_fe["Overtime_LowSatisfaction"],
    df_fe["Attrition"],
    normalize="index"
) * 100
```

```python
attrition_rate = (
    df_fe.groupby("Overtime_LowSatisfaction")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index()
)

attrition_rate["Overtime_LowSatisfaction"] = (
    attrition_rate["Overtime_LowSatisfaction"]
    .map({
        0: "Others",
        1: "Overtime + Low Satisfaction"
    })
)

sns.barplot(
    x="Overtime_LowSatisfaction",
    y="Attrition",
    data=attrition_rate
)

plt.title("Attrition Rate: Overtime + Low Satisfaction")
plt.xlabel("")
plt.ylabel("Attrition Rate (%)")

plt.show()
```
